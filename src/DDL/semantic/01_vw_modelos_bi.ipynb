## SEMANTICA: vista maestra de BI
## Oculta los joins del star schema. El negocio solo selecciona.

```python
%sql
CREATE OR REPLACE VIEW pf.semantic.vw_modelos_bi AS
SELECT
  dm.model_id AS modelo,
  dm.nombre AS nombre,
  dm.org_id AS organizacion,
  df.fecha AS fecha,
  df.anio AS anio,
  df.mes AS mes,
  dfc.fecha AS fecha_creacion,
  f.downloads AS descargas,
  f.likes AS likes,
  f.delta_downloads AS descargas_adicionales_dia,
  dt.pipeline_tag AS tarea,
  dlb.library_name AS libreria,
  dlc.license_tag AS licencia
FROM
  pf.gold.fact_metricas_diarias f
    JOIN pf.gold.dim_fecha df
      ON df.fecha_id = f.fecha_id
    JOIN pf.gold.dim_fecha dfc
      ON dfc.fecha_id = f.created_fecha_id
    JOIN pf.gold.dim_modelo_scd2 dm
      ON dm.modelo_id = f.modelo_id
      AND dm.is_current = TRUE
    JOIN pf.gold.dim_organizacion do
      ON do.org_sk = f.org_sk
    JOIN pf.gold.dim_task dt
      ON dt.task_id = f.task_id
    JOIN pf.gold.dim_libreria dlb
      ON dlb.libreria_id = f.libreria_id
    JOIN pf.gold.dim_licencia dlc
      ON dlc.licencia_id = f.licencia_id;
```

```python
%sql

SELECT * FROM pf.semantic.vw_modelos_bi ORDER BY descargas DESC LIMIT 10;
```
