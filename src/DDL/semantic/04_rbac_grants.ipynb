## GOBERNANZA: RBAC con GRANT (Unity Catalog)
## Ejecutar como account/metastore admin. Detalle en docs/05_gobernanza.md

```python
%sql
-- Acceso base al catalogo (usage privilege: no da acceso a datos por si solo)
GRANT USE CATALOG ON CATALOG pf TO `data-engineers`;

GRANT USE CATALOG ON CATALOG pf TO `data-analysts`;

GRANT USE CATALOG ON CATALOG pf TO `business-analysts`;

-- Ingenieros de datos: crear/escribir en todas las capas
-- Nota UC: CREATE es privilegio de CATALOGO (crear schemas); a nivel de schema
-- se conceden CREATE TABLE (cubre tablas Y vistas) / MODIFY / SELECT (+ USE SCHEMA).
GRANT USE SCHEMA ON SCHEMA pf.bronze TO `data-engineers`;

GRANT CREATE TABLE ON SCHEMA pf.bronze TO `data-engineers`;

GRANT MODIFY ON SCHEMA pf.bronze TO `data-engineers`;

GRANT SELECT ON SCHEMA pf.bronze TO `data-engineers`;

GRANT USE SCHEMA ON SCHEMA pf.silver TO `data-engineers`;

GRANT CREATE TABLE ON SCHEMA pf.silver TO `data-engineers`;

GRANT MODIFY ON SCHEMA pf.silver TO `data-engineers`;

GRANT SELECT ON SCHEMA pf.silver TO `data-engineers`;

GRANT USE SCHEMA ON SCHEMA pf.gold TO `data-engineers`;

GRANT CREATE TABLE ON SCHEMA pf.gold TO `data-engineers`;

GRANT MODIFY ON SCHEMA pf.gold TO `data-engineers`;

GRANT SELECT ON SCHEMA pf.gold TO `data-engineers`;

-- El pipeline crea las vistas semánticas (notebook 08) -> CREATE TABLE cubre vistas
GRANT USE SCHEMA ON SCHEMA pf.semantic TO `data-engineers`;

GRANT CREATE TABLE ON SCHEMA pf.semantic TO `data-engineers`;

-- Lectura del volumen (necesaria para el ETL de ingesta)
--GRANT READ VOLUME ON pf.landing.files TO `data-engineers`;
GRANT READ VOLUME ON VOLUME pf.landing.files TO `data-engineers`;

-- Analistas: lectura de silver y gold
GRANT USE SCHEMA ON SCHEMA pf.silver TO `data-analysts`;

GRANT SELECT ON SCHEMA pf.silver TO `data-analysts`;

GRANT USE SCHEMA ON SCHEMA pf.gold TO `data-analysts`;

GRANT SELECT ON SCHEMA pf.gold TO `data-analysts`;

-- Negocio: SOLO las vistas sematicas
GRANT USE SCHEMA ON SCHEMA pf.semantic TO `business-analysts`;

GRANT SELECT ON SCHEMA pf.semantic TO `business-analysts`;

-- (Opcional) leer una vista puntual
-- GRANT SELECT ON pf.semantic.vw_modelos_bi TO `business-analysts`;
```

```python
%sql
-- Verificar los permisos concedidos
SHOW GRANTS ON SCHEMA pf.bronze;
SHOW GRANTS ON SCHEMA pf.silver;
SHOW GRANTS ON SCHEMA pf.gold;
SHOW GRANTS ON SCHEMA pf.semantic;
```
