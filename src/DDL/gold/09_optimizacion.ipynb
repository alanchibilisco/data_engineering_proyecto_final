## Optimizacion de objetos (Gold) para acelerar el dashboard

```python
%sql
-- 1) Liquid Clustering en la tabla de hechos (filtros por modelo/fecha)
ALTER TABLE pf.gold.fact_metricas_diarias
  CLUSTER BY (model_id, fecha_id);

-- 2) OPTIMIZE (compacta y reordena segun clustering)
OPTIMIZE pf.gold.fact_metricas_diarias;

-- 3) Z-order en dim_modelo_scd2 para acelerar lookups de carga y consultas
OPTIMIZE pf.gold.dim_modelo_scd2 ZORDER BY (model_id, is_current);

-- 4) Estadisticas para el optimizador
ANALYZE TABLE pf.gold.fact_metricas_diarias COMPUTE STATISTICS;
ANALYZE TABLE pf.gold.dim_modelo_scd2 COMPUTE STATISTICS;
ANALYZE TABLE pf.gold.dim_fecha COMPUTE STATISTICS;
ANALYZE TABLE pf.gold.dim_organizacion COMPUTE STATISTICS;
ANALYZE TABLE pf.gold.dim_task COMPUTE STATISTICS;
ANALYZE TABLE pf.gold.dim_libreria COMPUTE STATISTICS;
ANALYZE TABLE pf.gold.dim_licencia COMPUTE STATISTICS;

-- 5) Mantenimiento (solo en ventana de full_refresh mensual)
-- VACUUM hf.gold.fact_metricas_diarias RETAIN 168 HOURS;
-- VACUUM hf.bronze.models_raw          RETAIN 168 HOURS;
```
