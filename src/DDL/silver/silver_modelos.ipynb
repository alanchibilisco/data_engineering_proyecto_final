```sql
%sql
-- =========================================================
-- DDL Silver: pf.silver.modelos
-- Limpio, tipado, deduplicado. Particion por ingestion_date.
-- =========================================================

DROP TABLE IF EXISTS pf.silver.modelos;

CREATE TABLE IF NOT EXISTS pf.silver.modelos (
    model_id        STRING    NOT NULL COMMENT 'BK - ID natural org/nombre',
    nombre          STRING    COMMENT 'Nombre corto (despues del /)',
    org_id          STRING    COMMENT 'Organizacion (antes del /)',
    likes           BIGINT    COMMENT 'Likes',
    downloads       BIGINT    COMMENT 'Descargas',
    private         BOOLEAN   COMMENT 'Repo privado',
    pipeline_tag    STRING    COMMENT 'Tarea (text-generation, ...)',
    library_name    STRING    COMMENT 'Libreria',
    license_tag     STRING    COMMENT 'Tag de licencia extraida (license:*)',
    createdAt       TIMESTAMP COMMENT 'Fecha de creacion tipada',
    lastModified    TIMESTAMP COMMENT 'Ultima modificacion tipada',
    ingestion_date  DATE      COMMENT 'Particion',
    has_metadata    BOOLEAN   COMMENT 'True si payload_json no es nulo (calidad)',
    _rescued_data   STRING    COMMENT 'Campos no mapeados'
)
USING DELTA
PARTITIONED BY (ingestion_date)
TBLPROPERTIES (
   delta.enableChangeDataFeed = true,
   delta.autoOptimize.optimizeWrite = true,
   delta.autoOptimize.autoCompact = true
)
COMMENT 'Silver: modelos limpios, tipados y deduplicados por (model_id, ingestion_date)';

```
