```sql
%sql
-- =========================================================
-- DDL Silver: pf.silver.model_tag
-- Tags explotados: una fila por (modelo, tag).
-- =========================================================

DROP TABLE IF EXISTS pf.silver.model_tag;

CREATE TABLE IF NOT EXISTS pf.silver.model_tag (
    model_id        STRING   COMMENT 'Modelo al que pertenece el tag',
    tag             STRING   COMMENT 'Texto del tag',
    es_license      BOOLEAN  COMMENT 'True si el tag es license:*',
    es_dataset      BOOLEAN  COMMENT 'True si el tag es dataset:*',
    es_arxiv        BOOLEAN  COMMENT 'True si el tag es arxiv:*',
    ingestion_date  DATE     COMMENT 'Particion'
)
USING DELTA
PARTITIONED BY (ingestion_date)
TBLPROPERTIES (
   delta.enableChangeDataFeed = true,
   delta.autoOptimize.optimizeWrite = true,
   delta.autoOptimize.autoCompact = true
)
COMMENT 'Silver: tags de modelos explotados (modelo x tag)';
```
